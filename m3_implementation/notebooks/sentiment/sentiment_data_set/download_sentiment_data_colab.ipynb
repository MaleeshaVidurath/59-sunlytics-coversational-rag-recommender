# Download SemEval 2017 Tweet Sentiment Dataset
### Sunlytics CRS — M3 Sentiment Classifier

Downloads the `cardiffnlp/tweet_eval` sentiment dataset from HuggingFace and saves train, validation, and test splits as separate CSV files.

**Output files:**
- `tweet_eval_train.csv` — 45,615 rows
- `tweet_eval_val.csv` — 2,000 rows
- `tweet_eval_test.csv` — 12,284 rows

**CSV columns:** `text`, `label`, `label_name`

| label | label_name |
|-------|------------|
| 0 | negative |
| 1 | neutral |
| 2 | positive |

```python
# ── CELL 2 — Load dataset from HuggingFace ────────────────────────────────────
from datasets import load_dataset
from collections import Counter

dataset = load_dataset('cardiffnlp/tweet_eval', 'sentiment')

ID2LABEL = {0: 'negative', 1: 'neutral', 2: 'positive'}

print('=== Dataset loaded ===')
for split in ['train', 'validation', 'test']:
    counts = Counter(dataset[split]['label'])
    print(f'\n{split.upper()} — {len(dataset[split]):,} rows')
    for lid, name in ID2LABEL.items():
        pct = counts[lid] / len(dataset[split]) * 100
        print(f'  {name:10s}: {counts[lid]:,}  ({pct:.1f}%)')
```

```python
# ── CELL 3 — Convert to DataFrames and save as CSV ────────────────────────────
import pandas as pd
import os

os.makedirs('data', exist_ok=True)

split_map = {
    'train':      'tweet_eval_train.csv',
    'validation': 'tweet_eval_val.csv',
    'test':       'tweet_eval_test.csv',
}

for split, filename in split_map.items():
    df = pd.DataFrame({
        'text':       dataset[split]['text'],
        'label':      dataset[split]['label'],
        'label_name': [ID2LABEL[l] for l in dataset[split]['label']],
    })
    path = f'data/{filename}'
    df.to_csv(path, index=False)
    print(f'Saved {path}  ({len(df):,} rows)')

print('\nSample rows from train:')
print(pd.read_csv('data/tweet_eval_train.csv').head(3).to_string())
```

```python
# ── CELL 4 — Download CSV files ───────────────────────────────────────────────
from google.colab import files

for filename in ['tweet_eval_train.csv', 'tweet_eval_val.csv', 'tweet_eval_test.csv']:
    files.download(f'data/{filename}')
    print(f'Downloading: {filename}')

print('\nAll downloads started.')
```
